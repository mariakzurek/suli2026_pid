```python
cols = ["pid", "p", "theta", "beta", "chi2pid", "rich_RQ", "vz", "bdt_pass", "rich_best_PID", "rich_RQ", "rich_best_ntot", "bdt_score"]
kinematics =["Mx_eKX","Mx_epiX","Mx_epX", "Q2", "W", "y"]

for kin in kinematics:
    cols.append(kin)
    
df = uproot.open("~/ML_Files/epkx_data/scored_large/epkx_dataset_large.root:PhysicsEvents").arrays(cols, library="pd")
df=df[df["rich_best_ntot"]>2]
df=df[df["rich_RQ"]>0.1]
df=au.apply_Sidis_Cuts(df)

outDir="../../figures/Data_Application/"

pEdges = au.makeBinEdges(3, 5, 10)

# Select RICH angular range
df_rich = df[df["theta"] < 20]
df_rich = df[df["p"] > 3]
# = df_rich[
 #   (df_rich["rich_best_PID"] == 321) |
  #  (df_rich["rich_best_PID"] == 221) |
   # (df_rich["rich_best_PID"] == 2212)
#]
df_rich = df_rich[
    (df_rich["rich_best_PID"]!=-9999)
]

denominator_df = df_rich[
        (df_rich["pid"] == 321) &
        (df_rich["bdt_pass"] == True)
    ]
denominator_df=au.apply_Sidis_Cuts(denominator_df)
```

```python
import matplotlib.pyplot as plt

# Column to plot
column = "rich_best_PID"

plt.figure(figsize=(8, 6))
plt.hist(denominator_df[column], bins=100)

plt.xlabel(column)
plt.ylabel("Counts")
plt.title(f"Histogram of {column}")
plt.yscale("log")
plt.tight_layout()
plt.savefig("rich_pid_log.png", dpi=150)
plt.show()
```
