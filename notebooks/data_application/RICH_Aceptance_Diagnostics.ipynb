```python
def apply_RICH_Cuts(df):
    df_out= df[df["p"]>=2.5]
    df_out= df_out[df_out["theta"]<=20]
    df_out= df_out[df_out["rich_best_ntot"]>=2.5]
    df_out= df_out[df_out["rich_best_ntot"]>=2.5]
    df_out= df_out[df_out["rich_best_PID"]!=-9999]
    df_out= df_out[df_out["rich_RQ"]>0.1]
    return df_out
```

```python
cols = ["pid", "p", "theta", "beta", "chi2pid", "rich_RQ", "vz", "bdt_pass", "rich_best_PID", "rich_RQ", "rich_best_ntot", "bdt_score"]
kinematics =["Mx_eKX","Mx_epiX","Mx_epX", "Q2", "W", "y"]

for kin in kinematics:
    cols.append(kin)
    
#df_dat = uproot.open("~/ML_Files/epkx_data/scored_larger/larger.root:PhysicsEvents").arrays(cols, library="pd")
df_dat = uproot.open("~/ML_Files/scored_data_v01/nSidis_005046.root:PhysicsEvents").arrays(cols, library="pd")
df_dat=df_dat[df_dat["rich_best_PID"]==321]

df_mc = uproot.open("~/ML_Files/MC_scored/pid_training_v2.root:PhysicsEvents").arrays(cols, library="pd")
df_mc=df_mc[df_mc["rich_best_PID"]==321]
outDir="../../figures/Data_Application/EB_Kaon/"

allPlots=[]
```

```python
edge=au.makeBinEdges(0,30,360)
```

```python
thetabins= au.makeBins(df_dat, "theta", binEdges=edge)
efficiency = []

for theta_bin in thetabins:
    cut_RICH = ((theta_bin["rich_best_PID"]!=-9999)&(theta_bin["rich_best_ntot"]>3)&(theta_bin["rich_RQ"]>0.2))
    before = len(theta_bin)

    # Apply the original mask to this theta bin
    after = len(theta_bin[cut_RICH])

    if before > 0:
        efficiency.append(after / before)
    else:
        efficiency.append(0)

# Plot
bin_centers = (edge[:-1] + edge[1:]) / 2

plt.figure(figsize=(8,6))
plt.plot(bin_centers, efficiency, 'o')
plt.xlabel(r'$\theta$ (deg)')
plt.ylabel('acceptance (Cut/pre-cut)')
plt.title("Acceptance vs theta RICH-K (DATA)")
plt.ylim(0, 0.2)
plt.grid(False)
plt.savefig("K_data_acceptance_vs_theta_RICH.png", dpi=150)
allPlots.append(plt.gcf())
plt.close()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Controls whether acceptance values are written in each square
SHOW_VALUES = True

# Define bins
p_edges = np.linspace(0.5, 5.0, 11)       # 10 p bins
theta_edges = np.linspace(0, 20, 11)      # 10 theta bins

# Store acceptance values
acceptance = np.zeros((len(theta_edges)-1, len(p_edges)-1))

# Loop through theta bins
theta_bins = au.makeBins(df_dat, "theta", binEdges=theta_edges)

for i, theta_bin in enumerate(theta_bins):

    # Now split this theta bin into momentum bins
    p_bins = au.makeBins(theta_bin, "p", binEdges=p_edges)

    for j, p_bin in enumerate(p_bins):

        # Define RICH cut
        cut_RICH = (
            (p_bin["rich_best_PID"] != -9999) &
            (p_bin["rich_best_ntot"] > 3) &
            (p_bin["rich_RQ"] > 0.2)
        )

        before = len(p_bin)
        after = len(p_bin[cut_RICH])

        if before > 0:
            acceptance[i, j] = after / before
        else:
            acceptance[i, j] = 0


# -----------------------------
# Plot heatmap
# -----------------------------

plt.figure(figsize=(9,7))

plt.imshow(
    acceptance,
    origin="lower",
    aspect="auto",
    extent=[
        p_edges[0],
        p_edges[-1],
        theta_edges[0],
        theta_edges[-1]
    ],
    vmin=0,
    vmax=1,
    cmap='coolwarm'
)

plt.colorbar(label="Acceptance (Cut / Pre-cut)")

plt.xlabel(r"$p$ (GeV/c)")
plt.ylabel(r"$\theta$ (deg)")
plt.title("RICH Acceptance vs p and theta RICH-K (DATA)")


# Add values to squares
if SHOW_VALUES:
    for i in range(len(theta_edges)-1):
        for j in range(len(p_edges)-1):

            p_center = (p_edges[j] + p_edges[j+1]) / 2
            theta_center = (theta_edges[i] + theta_edges[i+1]) / 2

            plt.text(
                p_center,
                theta_center,
                f"{acceptance[i,j]:.3f}",
                ha="center",
                va="center",
                fontsize=7
            )


plt.savefig("K_data_acceptance_2D_RICH.png", dpi=150, bbox_inches="tight")
allPlots.append(plt.gcf())
plt.close()
```

```python
from matplotlib.backends.backend_pdf import PdfPages
from PIL import Image
import matplotlib.pyplot as plt
import io

with PdfPages("torus-1_rich_acceptance.pdf") as pdf:

    for i in range(0, len(allPlots), 2):

        fig, axes = plt.subplots(1, 2, figsize=(16, 8))

        for ax, plot in zip(axes, allPlots[i:i+2]):

            # Render the existing figure
            buf = io.BytesIO()
            plot.savefig(buf, format="png", dpi=150)
            buf.seek(0)

            img = Image.open(buf)

            ax.imshow(img)
            ax.axis("off")

        # Hide unused panel if odd number of plots
        if len(allPlots[i:i+2]) == 1:
            axes[1].axis("off")

        fig.tight_layout()
        pdf.savefig(fig)
        plt.close(fig)
```
