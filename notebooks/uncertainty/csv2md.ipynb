```python

import pandas as pd


import csv

input_file = "weighted_contamination_comparison_granular.csv"
output_file = "eff_misId_grid.md"

with open(input_file, "r", newline="") as csvfile:
    reader = csv.reader(csvfile)
    rows = list(reader)

with open(output_file, "w") as mdfile:
    # Header
    mdfile.write("| " + " | ".join(rows[0]) + " |\n")

    # Separator
    mdfile.write("| " + " | ".join(["---"] * len(rows[0])) + " |\n")

    # Data rows
    for row in rows[1:]:
        mdfile.write("| " + " | ".join(row) + " |\n")
```
