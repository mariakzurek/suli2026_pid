```python
import matplotlib.pyplot as plt
import numpy as np
import uproot
import pandas as pd
import matplotlib.colors as colors
#open root file
df = uproot.open(
    "/volatile/clas12/cooperb/SULI/pid_training_test.root:PhysicsEvents"
).arrays(library="pd")
```

```python
#plots with no cuts
fig, ax = plt.subplots(figsize=(8, 6))

ax.hist2d(df["p"], df["beta"],
    bins=[80, 80], range=[[0.5, 5.0], [0.6, 1.1]],
    alpha=0.6, cmap="Reds",norm=colors.LogNorm())
ax.set_xlabel("p (GeV/c)")
ax.set_ylabel("beta")
ax.set_title("NOCUT beta vs p by MC truth class")
fig.tight_layout()
fig.savefig("figures/NOCUT_beta_vs_p_truth_classes.png", dpi=150)
plt.close(fig)
```

```python
#creates a list of matching events (excludes events that failed matching)
matched = df[df["mc_matching_pid"] != -9999].copy()
```

```python
#plots only pions
fig, ax = plt.subplots(figsize=(8, 6)) 
subset = matched[matched["mc_matching_pid"] == 211] #ask about what this line does
ax.hist2d(subset["p"], subset["beta"],
    bins=[80, 80], range=[[0.5, 5.0], [0.6, 1.1]],
    alpha=0.6, cmap="Blues")
ax.set_xlabel("p (GeV/c)")
ax.set_ylabel("beta")
ax.set_title("PI Only beta vs p by MC truth class")
fig.tight_layout()
fig.savefig("figures/PI_beta_vs_p_truth_classes.png", dpi=150)
plt.close(fig)
```

```python
#plots only kaons
fig, ax = plt.subplots(figsize=(8, 6))
subset = matched[matched["mc_matching_pid"] == 321] #ask about what this line does
ax.hist2d(subset["p"], subset["beta"],
    bins=[80, 80], range=[[0.5, 5.0], [0.6, 1.1]],
    alpha=0.6, cmap="Greens")
ax.set_xlabel("p (GeV/c)")
ax.set_ylabel("beta")
ax.set_title("K Only beta vs p by MC truth class")
fig.tight_layout()
fig.savefig("figures/K_beta_vs_p_truth_classes.png", dpi=150)
plt.close(fig)
```

```python
#plots only protons
fig, ax = plt.subplots(figsize=(8, 6))
subset = matched[matched["mc_matching_pid"] == 2212] #ask about what this line does
ax.hist2d(subset["p"], subset["beta"],
    bins=[80, 80], range=[[0.5, 5.0], [0.6, 1.1]],
    alpha=0.6, cmap="Reds")
ax.set_xlabel("p (GeV/c)")
ax.set_ylabel("beta")
ax.set_title("P Only beta vs p by MC truth class")
fig.tight_layout()
fig.savefig("figures/P_beta_vs_p_truth_classes.png", dpi=150)
plt.close(fig)
```

```python
#plots without Protons only
fig, ax = plt.subplots(figsize=(8, 6))
for pid_val, label, color in [(211, "pi+", "blue"),
                               (321, "K+",  "green")]:
    subset = matched[matched["mc_matching_pid"] == pid_val] #ask about what this line does
    ax.hist2d(subset["p"], subset["beta"],
              bins=[50, 50], range=[[0.5, 5.0], [0.6, 1.1]],
              alpha=0.6, label=label, cmap="Blues" if pid_val==211 else
                                          "Greens" if pid_val==321 else "Reds")
ax.set_xlabel("p (GeV/c)")
ax.set_ylabel("beta")
ax.set_title("No P beta vs p by MC truth class")
fig.tight_layout()
fig.savefig("figures/NOP_beta_vs_p_truth_classes.png", dpi=150)
plt.close(fig)
```
